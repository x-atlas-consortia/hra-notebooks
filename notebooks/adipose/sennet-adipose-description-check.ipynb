## Code to get the description for adipose datasets in SenNet

## Install and import libraries

```python
%pip install requests pandas

import requests
from pprint import pprint
import json
import pandas as pd
```

## Define functions

```python
BEARER_TOKEN = "" # fill this in to get unpublished data as well


def make_api_call(id: str) -> dict:
    BASE_URL = "https://entity.api.sennetconsortium.org/entities/"

    headers = {
        "Authorization": f"Bearer {BEARER_TOKEN}",
        "Content-Type": "application/json",
    }

    response = requests.get(BASE_URL + id, headers = headers)

    print(response.status_code)
    # print(response.text)
    
    data = response.json()
    return data
```

## Load dataset IDs and run query

```python
with open("sennet-adipose-ids.json") as f:
    data = json.load(f)

uuid_list = [item["id"] for item in data]
print(uuid_list)
```

```python
max = len(uuid_list)
result = pd.DataFrame()
dataset_id = []
sennet_id = []
types = []

for i in range(max):

    if i < max:
        print(f"Now at loop {i}:")

        response = make_api_call(uuid_list[i])
        try:
            description = response["direct_ancestor"]["description"]
        except KeyError:
            description = [item for item in response["origin_samples"]][0]["description"]
        # finally: 
        #     description = "no description in API response"

        print(description)
        dataset_id.append(uuid_list[i])
        sennet_id.append(response["direct_ancestor"]["sennet_id"])
        types.append(description)

        i = i + 1
        print()

result[["dataset_id", "uuid", "type"]] = pd.DataFrame(
    {
        "dataset_id": dataset_id,
        "uuid": sennet_id,
        "type": types,
    }
)

print(result)
```

```python
# Export result
result.to_csv("sennet_adipose.csv", index=False)
```
